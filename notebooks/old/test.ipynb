```python
from wtlike import WtLike
from utilities.ipynb_docgen import capture_hide

def display_source(row, ax):
    with capture_hide():
        name, info = row
        wtl = WtLike(name)
        wtl.view(365.25).bb_view().plot(ax=ax)
    ax.get_legend().remove()
    ax.set(ylabel='', xlabel='')
    ax.axvline(wtl.start+7*info.time, color='red', ls='--', lw=2)
    return info

def show_multiplots(df: pd.DataFrame, 
            display_source: callable, 
            ncols:int=3, 
            subfigsize:tuple=(4,3),
            caption:str='',
            title:str=''):
    
    n = len(df)
    nrows = (n+ ncols-1)//ncols
    fig, axx = plt.subplots(ncols=ncols, nrows=nrows, 
                        figsize=np.array(subfigsize)*np.array([ncols,nrows]), 
                        gridspec_kw=dict(top=0.95, left=0.05, bottom=0.05, right=0.95),
                        sharex=True, sharey=True)
    tooltips = []
    for row, ax in zip(df.iterrows(), axx.flat):
        tooltips.append(display_source(row, ax))
    for ax in axx.flat[n:]:
        ax.set_visible(False)
    if not caption: 
        caption=f"""This map has tooltips. Scales for x and y axes are {ax.get_xlim()}
         and {ax.get_ylim()}."""
    if title:
        fig.suptitle(title)
    fig.facecolor='0.1'
    show(fig, 
         tooltips=tooltips,
         caption=caption)   
```

```python
show_multiplots(bdfq[:21].sort_index(), display_source, 
                subfigsize=(5,2), ncols=3)      
```
